**While loop**

- If you want to appply unlimited iterations then we choose while loop

- Any loop we required 3 things

    - Intial point or start point

    - Condition to stop

    - Increment or decrement

- In for loop we are using all these 3 in a single line

- In while Loop we will use all these 3 in 3 lines



```python
# intial point
# While <condition>:
#     <statments>
#     <increament or decrement>
```

```python
for i in range(1,6):
    print(i,end=' ')

# i=1
# when i reaches 5 then should stop 
# i=i+1
```

```python
i=1
while <condition>:
    print(i,end=' ')
    i=i+1

# Condition anything that should be True
# Then you will go inside the while loop
# Ansd execute

i=1
i<=6
i<2
i<3
i<4
```

```python
# Case-1:
i=1   # Start point
while i<3:  # Condition
    print(i,end=' ')
    i=i+1   # Increment or decrement

# step=1 :  i=1  while 1<3  True  print(1)  i=1+1=2
# step-2:  i=2   while 2<3  True  print(2)  i=1+2=3
# ste-3:  i=3   while 3<3  False
```

```python
i=1
while(i<6):
    print(i,end=' ')
    i=i+1
```

**Infinite loop**

- whenever infinite loop occures

- apply the condition and then apply the break

```python
i=1
while i!=0:
    print(i,end=' ')
    i=i+1
# step-1
```

```python
i=1
while i!=0:
    print(i,end=' ')
    i=i+1
    if i==10:
        break
```

```python
i=1
while i<20:
    print(i)
    i=i/2
# i=1  1<20 T   print(1)  1=1/2
```

**True**

- When we want to use infinite iterations will use True

- If we want to stop the loop, inside we will apply a condition and break

```python
i=1
while True:
    print(i,end=' ')
    i=i+1
    if i==10:
        break
```
